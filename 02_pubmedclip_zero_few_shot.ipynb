````python
from transformers import CLIPProcessor, CLIPModel

model_name = "flaviagiammarino/pubmed-clip-vit-base-patch32"

processor = CLIPProcessor.from_pretrained(model_name)
model = CLIPModel.from_pretrained(model_name)
model = model.to(device)

print("Model loaded!")
````

7. データローダー作成

````python
def clip_transform(image):
    return processor(images=image, return_tensors="pt")["pixel_values"].squeeze(0)

dataset = datasets.ImageFolder(data_dir, transform=clip_transform)

loader = DataLoader(
    dataset,
    shuffle=False,
    batch_size=val_batch_size,
    num_workers=num_workers
)

classes = dataset.classes

print(classes)
````

8. Prompt を作成

````python
# ===============================================
# Prompt Engineering
# ===============================================
text_prompts = [
    ["this is an image of " + cls for cls in classes],
    [cls + " presented in image" for cls in classes],
    ["a photo of " + cls for cls in classes],
]
````

9. Zero-shot 推論関数

````python
def evaluate(model, dataloader, text_tokens):
    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        text_features = model.get_text_features(**text_tokens)
        text_features = F.normalize(text_features, dim=-1)

        for images, labels in dataloader:
            images = images.to(device)
            labels = labels.to(device)

            image_features = model.get_image_features(pixel_values=images)
            image_features = F.normalize(image_features, dim=-1)

            logits = model.logit_scale.exp() * image_features @ text_features.T
            preds = logits.argmax(dim=1)

            correct += (preds == labels).sum().item()
            total += labels.size(0)

    return correct / total
````

10. Zero-shot 分類の実行

````python
for prompt in text_prompts:
    text_tokens = processor(
        text=prompt, return_tensors="pt", padding=True
        ).to(device)

    acc = evaluate(model, loader, text_tokens)

    print(prompt)
    print(f"Accuracy: {acc:.2%}")
````

11. Few-shot Dataset

````python
# ==================================================
# {object}専用 Dataset
# ==================================================
class Dataset(Dataset):
    def __init__(self, root_dir, object="BUSI", split="train", transform=None):
        """
        root_dir:
            {object}/
              ├── {object}/
              └── split_{object}.json

        split:
            train / val / test
        """
        self.root_dir = root_dir
        self.image_root = os.path.join(root_dir, object)
        self.transform = transform

        json_path = os.path.join(root_dir, f"split_{object}.json")

        with open(json_path, "r", encoding="utf-8") as f:
            split_info = json.load(f)

        self.samples = split_info[split]

        # class名取得
        self.classes = sorted(list(set([x[2] for x in self.samples])))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        rel_path, label, class_name = self.samples[idx]

        img_path = os.path.join(self.image_root, rel_path)
        image = Image.open(img_path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, label

# ==================================================
def get_subset_datasets(dataset, samples_per_class):
    """
    各クラスから指定枚数ランダム抽出
    """
    class_indices = defaultdict(list)

    for idx, sample in enumerate(dataset.samples):
        label = sample[1]
        class_indices[label].append(idx)

    selected_indices = []

    for indices in class_indices.values():
        selected = random.sample(
            indices,
            k=min(samples_per_class, len(indices))
        )
        selected_indices.extend(selected)

    return Subset(dataset, selected_indices)

# ==================================================
# Train Loader
# ==================================================
def creat_train_loader(
    root_folder,
    transform=None,
    object="BUSI",
    shot_num=1,
    num_workers=0
):

    train_dataset = Dataset(
        root_dir=root_folder,
        object=object,
        split="train",
        transform=transform
    )

    train_subset = get_subset_datasets(train_dataset, shot_num)

    num_classes = len(train_dataset.classes)
    train_batch_size = shot_num * num_classes

    print(f"Training datasets size: {len(train_subset)}")
    print(f"Number of classes: {num_classes}")
    print(f"train_batch_size: {train_batch_size}")

    train_loader = DataLoader(
        train_subset,
        batch_size=train_batch_size,
        shuffle=True,
        pin_memory=True,
        num_workers=num_workers,
        persistent_workers=num_workers > 0
    )

    return train_loader


# ==================================================
# Val/Test Loader
# ==================================================
def creat_val_test_loader(root_folder,
                          object="BUSI",
                          transform=None,
                          val_batch_size=256,
                          test_batch_size=256,
                          num_workers=8):

    val_dataset = Dataset(
        root_dir=root_folder,
        object=object,
        split="val",
        transform=transform
    )

    test_dataset = Dataset(
        root_dir=root_folder,
        object=object,
        split="test",
        transform=transform
    )

    valid_loader = DataLoader(
        val_dataset,
        batch_size=val_batch_size,
        shuffle=False,
        pin_memory=True,
        num_workers=num_workers,
        persistent_workers=num_workers > 0
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=test_batch_size,
        shuffle=False,
        pin_memory=True,
        num_workers=num_workers,
        persistent_workers=num_workers > 0
    )

    print(f"Object: {object}")
    print(f"Val size: {len(val_dataset)}")
    print(f"Test size: {len(test_dataset)}")
    print(f"Number of classes: {len(val_dataset.classes)}")

    return valid_loader, test_loader, val_dataset.classes
````

12. Linear Probe Model

````python
class LinearProbe(nn.Module):
    def __init__(self, clip_model, num_classes):
        super().__init__()
        self.clip_model = clip_model

        for param in self.clip_model.parameters():
            param.requires_grad = False

        self.classifier = nn.Linear(512, num_classes)

    def forward(self, x):
        features = self.clip_model.get_image_features(pixel_values=x)
        features = F.normalize(features, dim=-1)
        logits = self.classifier(features)
        return logits
````

13. Train Function

````python
def train(
    model,
    loader,
    optimizer,
    criterion
):

    model.train()

    total_loss = 0

    for images, labels in loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        logits = model(images)

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    return total_loss / len(loader)
````

Few-shot 推論関数

````python
def evaluate_few_shot(
    model,
    loader
):

    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)
            labels = labels.to(device)

            logits = model(images)

            preds = logits.argmax(1)

            correct += (
                preds == labels
            ).sum().item()

            total += labels.size(0)

    return correct / total
````

14. Few-shot Experiment

````python
# ==========================================================
# Few-shot loop
# ==========================================================
for n in range(5):

    shot_num = 2 ** n

    print(f"\n========== {shot_num}-shot ==========")

    train_loader = creat_train_loader(
        root_folder="/content/drive/MyDrive//YOUR_USERNAME/BTMRI",
        transform=train_transform,
        object=OBJECT,
        shot_num=shot_num,
        num_workers=num_workers
    )

    LP_model = LinearProbe(
        model,
        len(classes)
    ).to(device)

    optimizer = torch.optim.Adam(
        LP_model.parameters(),
        lr=1e-3
    )

    criterion = nn.CrossEntropyLoss()

    for epoch in range(20):

        loss = train(
            LP_model,
            train_loader,
            optimizer,
            criterion
        )

    acc = evaluate_few_shot(
        LP_model,
        test_loader
    )

    print(
        f"Test Accuracy: {acc:.2%}"
    )
````
